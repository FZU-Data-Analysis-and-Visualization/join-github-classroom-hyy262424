```python
# 导入绘图库
import matplotlib.pyplot as plt
import numpy as np

# ===================== 1. 录入原始数据（和表格完全一致） =====================
departments = ["销售部", "技术部", "市场部", "财务部", "人事部"]
overtime_hours = [5032, 3614, 2204, 1025, 796]

# 计算累计百分比
total = sum(overtime_hours)
cumulative_percent = []
current_sum = 0
for hour in overtime_hours:
    current_sum += hour
    pct = current_sum / total * 100
    cumulative_percent.append(pct)

# ===================== 2. 创建画布、双坐标轴 =====================
# 设置中文显示（解决汉字乱码）
plt.rcParams["font.family"] = ["SimHei", "WenQuanYi Micro Hei", "Heiti TC"]
plt.rcParams["axes.unicode_minus"] = False

fig, ax1 = plt.subplots(figsize=(12, 7))  # 左坐标轴：柱状图
ax2 = ax1.twinx()  # 右坐标轴：折线图

# 绘制蓝色柱状图（加班时长）
bar_color = "#4477bb"
ax1.bar(departments, overtime_hours, color=bar_color, label="加班时长")
ax1.set_ylabel("加班时长", fontsize=12)
ax1.set_ylim(0, 6200)  # Y轴上限对齐原图6000刻度

# 绘制橙色折线（累计百分比）
line_color = "#ee7722"
ax2.plot(departments, cumulative_percent, color=line_color, linewidth=3, label="累计百分比")
ax2.set_ylabel("累计百分比", fontsize=12)
ax2.set_ylim(0, 120)  # 右侧百分比0~120%，匹配原图

# 设置右侧Y轴刻度（0%、20%...120%）
ax2.set_yticks([0, 20, 40, 60, 80, 100, 120])
ax2.set_yticklabels(["0.00%", "20.00%", "40.00%", "60.00%", "80.00%", "100.00%", "120.00%"])

# ===================== 3. 标题、图例、排版 =====================
plt.title("图表标题", fontsize=18, pad=20)

# 合并两个坐标轴的图例
handles1, labels1 = ax1.get_legend_handles_labels()
handles2, labels2 = ax2.get_legend_handles_labels()
ax1.legend(handles1 + handles2, labels1 + labels2, loc="lower center", bbox_to_anchor=(0.5, -0.15), ncol=2, frameon=False)

plt.tight_layout()  # 自动调整布局，防止文字被截断
plt.show()

# 打印核对百分比（和表格数据对照）
print("部门\t\t加班时长\t累计百分比")
for i in range(len(departments)):
    print(f"{departments[i]}\t{overtime_hours[i]}\t\t{cumulative_percent[i]:.2f}%")
```
