### 二、列与行操作

1. 选取 score 和 gender 两列。

   

2. 选取索引 0、2 的行。

   

3. 筛选 score>80 的行。

   

4. 新增一列 grade：score≥90 为优秀，否则为良好。

   

5. 删除 gender 列。

```python
import pandas as pd

data_dict = {
    'name': ['Tom', 'Jerry', 'Mike'],
    'score': [85, 92, 78],
    'gender': ['男', '女', '男']
}
df = pd.DataFrame(data_dict)

# 1. 选取score和gender两列
print(df[['score', 'gender']])

# 2. 选取索引0、2的行
print(df.loc[[0, 2]])

# 3. 筛选score>80的行
print(df[df['score'] > 80])

# 4. 新增grade列
df['grade'] = df['score'].apply(lambda x: '优秀' if x >= 90 else '良好')
print(df)

# 5. 删除gender列
print(df.drop('gender', axis=1))
```

### 三、缺失值

1. 创建含缺失值的 DataFrame：

   data = {"x":[1, None, 3], "y":[None, 5, 6]}

   检测缺失值并统计每列缺失数量。

   

2. 用 0 填充缺失值；删除含缺失值的行。

```python
import pandas as pd

# 1. 创建含缺失值的DataFrame并统计缺失数量
data_miss = {"x": [1, None, 3], "y": [None, 5, 6]}
df_miss = pd.DataFrame(data_miss)
print(df_miss.isnull())
print(df_miss.isnull().sum())

# 2. 填充缺失值和删除含缺失值的行
print(df_miss.fillna(0))
print(df_miss.dropna())
```

### 四、分组聚合

1. 按 gender 分组，求 score 的均值和最大值。

   

2. 按 gender 分组，统计每组人数。

```python
import pandas as pd

data_dict = {
    'name': ['Tom', 'Jerry', 'Mike'],
    'score': [85, 92, 78],
    'gender': ['男', '女', '男']
}
df = pd.DataFrame(data_dict)

# 1. 按gender分组求score的均值和最大值
print(df.groupby('gender')['score'].agg(['mean', 'max']))

# 2. 按gender分组统计每组人数
print(df.groupby('gender').size())
```

### 五、排序与去重

1. 按 score 升序排序。

   

2. 给 DataFrame 添加一行重复数据（如 Tom,85, 男），去重。


```python
import pandas as pd

data_dict = {
    'name': ['Tom', 'Jerry', 'Mike'],
    'score': [85, 92, 78],
    'gender': ['男', '女', '男']
}
df = pd.DataFrame(data_dict)

# 1. 按 score 升序排序
print(df.sort_values('score', ascending=True))

# 2. 添加重复数据并去重（用 pd.concat 替代 append）
new_row = pd.DataFrame([{'name': 'Tom', 'score': 85, 'gender': '男'}])
df = pd.concat([df, new_row], ignore_index=True)
print(df.drop_duplicates())
```

### 六、合并与拼接

1. 横向合并：

   df1 = pd.DataFrame ({"id":[1,2], "name":["X","Y"]})

   df2 = pd.DataFrame ({"id":[1,2], "age":[20,30]})

   按 id 合并。

   

2. 纵向拼接：

   df_a = pd.DataFrame ({"val":[1,2]})

   df_b = pd.DataFrame ({"val":[3,4]})

   拼接后重置索引。

```python
import pandas as pd

# 1. 横向合并
df1 = pd.DataFrame({"id": [1, 2], "name": ["X", "Y"]})
df2 = pd.DataFrame({"id": [1, 2], "age": [20, 30]})
print(pd.merge(df1, df2, on='id'))

# 2. 纵向拼接
df_a = pd.DataFrame({"val": [1, 2]})
df_b = pd.DataFrame({"val": [3, 4]})
print(pd.concat([df_a, df_b], ignore_index=True))
```

### 七、时间处理

1. 创建时间列：["2025-01-01","2025-01-02","2025-01-03"]，转为 datetime 格式。

   

2. 提取时间列的月份、星期。

```python
import pandas as pd

# 1. 转为datetime格式
time_list = ["2025-01-01", "2025-01-02", "2025-01-03"]
df_time = pd.DataFrame({"date": time_list})
df_time['date'] = pd.to_datetime(df_time['date'])
print(df_time)

# 2. 提取月份、星期
df_time['month'] = df_time['date'].dt.month
df_time['weekday'] = df_time['date'].dt.weekday
print(df_time)
```

### 八、高级操作

1. 对 score 列做标准化（(值 - 均值)/ 标准差）。

   

2. 用透视表：行是 gender，值是 score，聚合函数为中位数。

```python
import pandas as pd

data_dict = {
    'name': ['Tom', 'Jerry', 'Mike'],
    'score': [85, 92, 78],
    'gender': ['男', '女', '男']
}
df = pd.DataFrame(data_dict)

# 1. 对score列做标准化
df['score_std'] = (df['score'] - df['score'].mean()) / df['score'].std()
print(df[['score', 'score_std']])

# 2. 透视表
print(df.pivot_table(values='score', index='gender', aggfunc='median'))
```
